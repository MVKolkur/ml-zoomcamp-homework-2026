## Q7 Sum of weights

```python
asia = df[df['origin'] == 'Asia']
```

```python
X = asia[['vehicle_weight', 'model_year']]
```

```python
X = X.head(7)
```

```python
X
```

```python
X = X.to_numpy()
```

```python
X
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
temp = XTX_inv.dot(X.T)
w = temp.dot(y)
```

```python
w.sum()
```
